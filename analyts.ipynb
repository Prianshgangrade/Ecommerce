```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("transactions.csv")

```

```python
df.info()
```

```python
df.head()
```

```python
df.describe()
```

```python
df["order_datetime"] = pd.to_datetime(df["order_datetime"])
```

```python
user_id = 1002
user_df = df[df["user_id"] == user_id]
user_df["line_total"].sum()
user_df.groupby("category_name")["line_total"].sum().idxmax()
```

```python
user_df
```

```python
df.groupby("product_name")["quantity"].sum().sort_values(ascending=False)
```

```python
df.groupby("category_name")["line_total"].sum().sort_values(ascending=False)
```

```python
df.groupby("user_id")["line_total"].sum().sort_values(ascending=False)
```

```python
spending_trend = user_df.groupby(user_df["order_datetime"].dt.date)["line_total"].sum().reset_index()
sns.lineplot(data=spending_trend,x="order_datetime",y="line_total",marker="o")
plt.title(f"User {user_id} Spending Trend")
plt.show()
```

```python
user_category = user_df.groupby("category_name")["line_total"].sum().reset_index()
sns.barplot(data=user_category,x="line_total",y="category_name")
plt.title("User Category Spending")
plt.show()
```

```python
product_sales = df.groupby("product_name")["quantity"].sum().sort_values(ascending=False).reset_index()
sns.barplot(data=product_sales,x="quantity",y="product_name")
plt.title("Best Selling Products")
plt.show()
```

```python
category_revenue = df.groupby("category_name")["line_total"].sum().sort_values(ascending=False).reset_index()
sns.barplot(data=category_revenue,x="line_total",y="category_name")
plt.title("Category Revenue")
plt.show()
```

```python
daily_revenue = df.groupby(df["order_datetime"].dt.date)["line_total"].sum().reset_index()
sns.lineplot(data=daily_revenue,x="order_datetime",y="line_total",marker="o")
plt.title("Overall Revenue Trend")
plt.show()
```
